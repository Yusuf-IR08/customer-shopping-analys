```python
summary = pd.DataFrame({
    "Metric": [
        "Total Transaksi",
        "Total Amount",
        "Rata rata Amount per Transaksi",
        "Jumlah Location",
        "Jumlah Kategori Produk",
        "Jumlah Size",
        "Jumlah Pilihan Warna",
        "Jumlah Jenis Pengiriman",
        "Jumlah Payment"
    ],
    "Value": [
        df["Customer ID"].count(),
        df["Purchase Amount (USD)"].sum(),
        df["Purchase Amount (USD)"].mean(),
        df["Location"].nunique(),
        df["Category"].nunique(),
        df["Size"].nunique(),
        df["Color"].nunique(),
        df["Shipping Type"].nunique(),
        df["Payment Method"].nunique()
    ]
})

summary["Value"] = summary["Value"].apply(
    lambda x: round(x, 2) if isinstance(x, (int, float)) else x
)

summary
```

## Amount Analysis

```python
df["Purchase Amount (USD)"].describe()
```

```python
fig = px.histogram(
    df["Purchase Amount (USD)"],
    nbins=50
)

fig.update_traces(
    marker_line_width=1,
    marker_line_color="black"
)

fig.show()
```

```python
fig = px.box(
    df["Purchase Amount (USD)"]
)

fig.show()
```

**Insight:**
- Berdasarkan histogram, distribusi nilai Purchase Amount terlihat cukup merata pada rentang 20–100 USD tanpa adanya lonjakan ekstrem pada nilai tertentu.
- Hal ini juga didukung oleh boxplot yang menunjukkan tidak adanya outlier, sehingga distribusi transaksi dapat dikatakan relatif stabil dan tidak didominasi oleh transaksi dengan nilai yang terlalu tinggi maupun terlalu rendah.
- Selain itu, median amount berada di sekitar 60 USD, dengan 50% transaksi berada pada rentang 39–81 USD, menunjukkan bahwa sebagian besar customer melakukan transaksi pada nilai pembelian menengah.
